# Speech Processing – VAD, MFCC, Visualization

```python
!pip install librosa

import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# Load example speech
y, sr = librosa.load(librosa.example('libri1'), duration=5)

# Plot waveform
plt.figure(figsize=(10, 3))
librosa.display.waveshow(y, sr=sr)
plt.title('Waveform')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.show()

# Compute MFCCs
mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)

# Display MFCCs
plt.figure(figsize=(10, 4))
librosa.display.specshow(mfccs, x_axis='time')
plt.colorbar()
plt.title('MFCC')
plt.tight_layout()
plt.show()
```
