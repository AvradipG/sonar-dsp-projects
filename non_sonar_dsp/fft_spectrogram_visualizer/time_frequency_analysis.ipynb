# Time-Frequency Analysis using Gabor Transform

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Simulated signal
fs = 1000
t = np.linspace(0, 1, fs)
x = np.sin(2*np.pi*100*t) + np.sin(2*np.pi*200*t) + np.random.normal(0, 0.3, fs)

# FFT
f = np.fft.rfftfreq(len(t), 1/fs)
X = np.abs(np.fft.rfft(x))
plt.figure()
plt.plot(f, X)
plt.title('FFT')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.grid(True)
plt.show()

# Spectrogram (Gabor)
f, t_spec, Sxx = signal.spectrogram(x, fs, window=('gaussian', 50), nperseg=128, noverlap=100)
plt.figure()
plt.pcolormesh(t_spec, f, 10*np.log10(Sxx), shading='gouraud')
plt.title('Gabor Spectrogram')
plt.xlabel('Time (s)')
plt.ylabel('Frequency (Hz)')
plt.colorbar(label='Power/Frequency (dB/Hz)')
plt.tight_layout()
plt.show()
```
