# Wavelet Denoising
Using Discrete Wavelet Transform (DWT) to denoise signals and compare with traditional filters.

```python
import numpy as np
!pip install PyWavelets

import pywt
import matplotlib.pyplot as plt

fs = 1000
t = np.linspace(0, 1, fs)
clean = np.sin(2*np.pi*50*t)
noise = np.random.normal(0, 0.5, fs)
noisy = clean + noise

wavelet = 'db4'
coeffs = pywt.wavedec(noisy, wavelet, level=4)
threshold = 0.4
coeffs_thresh = [pywt.threshold(c, threshold, mode='soft') if i > 0 else c for i, c in enumerate(coeffs)]
denoised = pywt.waverec(coeffs_thresh, wavelet)

plt.figure(figsize=(12, 6))
plt.plot(t, noisy, label='Noisy', alpha=0.5)
plt.plot(t, clean, label='Clean')
plt.plot(t, denoised[:len(t)], label='Wavelet Denoised', linestyle='--')
plt.legend()
plt.grid(True)
plt.title('Wavelet Denoising vs Clean Signal')
plt.show()
```
